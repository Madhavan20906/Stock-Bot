# 05 — Portfolio Optimization & Analysis

Markowitz mean-variance optimization and risk analysis.

```python
import sys; sys.path.insert(0, '..')
import yaml

from src.data.loader import load_prices_panel
from src.portfolio.optimizer import PortfolioOptimizer
from src.portfolio.risk import full_risk_report
from src.backtesting.metrics import monthly_returns_table, underwater_periods
from src.visualization.plots import portfolio_performance, drawdown_chart, correlation_heatmap

with open('../configs/portfolio.yaml') as f:
    config = yaml.safe_load(f)
```

```python
# Load price panel
tickers = config['data']['tickers']
prices = load_prices_panel(tickers, start='2018-01-01', end='2023-12-31')
print(prices.shape)
prices.tail(3)
```

```python
# Correlation heatmap
correlation_heatmap(prices).show()
```

```python
# Optimize portfolio
opt = PortfolioOptimizer(prices)
weights = opt.max_sharpe()
perf = opt.performance_summary(weights)
print('Weights:', {k: f"{v:.1%}" for k, v in weights.items() if v > 0.001})
print('Performance:', perf)
```

```python
# Discrete allocation
alloc, leftover = opt.allocate(weights, total_portfolio_value=100_000)
print(f'Allocation: {alloc}')
print(f'Leftover cash: ${leftover:.2f}')
```
